```python
import warnings
warnings.filterwarnings('ignore', category=UserWarning, module='sklearn')

import os
os.environ["OMP_NUM_THREADS"] = "2"

import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# 1. Prepare your data (replace with your actual data variable)
# X = ... 

# 2. Define the range of clusters to try
max_k = 10
k_values = range(1, max_k + 1)
wcss = [] # This will be the list of y-values

# 3. Use a loop to compute WCSS for each k
for i in k_values:
    # Initialize KMeans with the current number of clusters
    kmeans = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=42)
    # Fit the model to the data
    kmeans.fit(X)
    # Append the calculated inertia (WCSS) to our list
    wcss.append(kmeans.inertia_)

# 4. Now that 'k_values' and 'wcss' have the same length (10), we can plot them
plt.figure(figsize=(10, 6))
plt.plot(k_values, wcss, marker='o', linestyle='--')
plt.title('Elbow Method For Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Within-Cluster Sum of Squares (WCSS)')
plt.grid(True)
plt.show()
```

```python
# 1. Import the required libraries
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# 2. Prepare the dataset
X = [[1, 2], [2, 3], [3, 3], [8, 8], [9, 8], [8, 9]]

# 3. Define the range of clusters
max_k = 5
k_values = range(1, max_k + 1)
wcss = []

# 4. Calculate WCSS for each K value
for i in k_values:
    kmeans = KMeans(n_clusters=i, init='k-means++',
                    max_iter=300, n_init=10, random_state=42)

    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

# 5. Plot the Elbow graph
plt.figure(figsize=(10, 6))
plt.plot(k_values, wcss, marker='o', linestyle='--')

plt.title('Elbow Method For Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.grid(True)
plt.show()
```

```python
# 1. Import the required libraries
from sklearn.metrics import confusion_matrix, accuracy_score

# 2. Prepare actual and predicted values
actual = [1, 0, 1, 1, 0, 0]
predicted = [1, 0, 1, 0, 0, 1]

# 3. Calculate the confusion matrix
cm = confusion_matrix(actual, predicted)

# 4. Calculate the accuracy score
accuracy = accuracy_score(actual, predicted)

# 5. Display the results
print("Confusion Matrix:")
print(cm)

print("Accuracy:", accuracy)
```

```python
# 1. Import the required libraries
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 2. Load the Iris dataset
data = load_iris()
X = data.data
y = data.target

# 3. Split the dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 4. Initialize two Decision Tree models
model1 = DecisionTreeClassifier(random_state=42)
model2 = DecisionTreeClassifier(max_depth=2, random_state=42)

# 5. Train both models
model1.fit(X_train, y_train)
model2.fit(X_train, y_train)

# 6. Calculate and display accuracy
print("Deep Tree Training Accuracy:",
      accuracy_score(y_train, model1.predict(X_train)))

print("Deep Tree Testing Accuracy:",
      accuracy_score(y_test, model1.predict(X_test)))

print("Limited Tree Training Accuracy:",
      accuracy_score(y_train, model2.predict(X_train)))

print("Limited Tree Testing Accuracy:",
      accuracy_score(y_test, model2.predict(X_test)))
```
